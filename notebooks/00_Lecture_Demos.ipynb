## 1️⃣ Tokens: the AI doesn't read words, it reads *pieces*
Before an AI reads your message, it chops it into **tokens**. Each token gets an ID number.
This is the same tokenizer used by GPT-4o.

```python
# @title 🔧 Load the tokenizer
import tiktoken
encoder = tiktoken.get_encoding("o200k_base")
COLORS = ["#ffadad", "#ffd6a5", "#fdffb6", "#caffbf", "#9bf6ff", "#a0c4ff", "#bdb2ff", "#ffc6ff"]


def show_tokens(text):
    ids = encoder.encode(text)
    pieces = [encoder.decode_single_token_bytes(i).decode("utf-8", errors="replace") for i in ids]
    boxes = "".join(
        f'<span style="background:{COLORS[n % len(COLORS)]};color:black;padding:3px 2px;'
        f'margin:1px;border-radius:4px;font-size:22px;font-family:monospace">'
        f'{html.escape(p).replace(" ", "&nbsp;")}</span>'
        for n, p in enumerate(pieces))
    display(HTML(f"<div style='margin:8px 0'>{boxes} &nbsp;&nbsp;<b style='font-size:18px'>{len(ids)} tokens</b></div>"))
    print("   Token IDs:", ids)
```

```python
show_tokens("Cisco routers connect the internet.")
show_tokens("Mój ulubiony sklep to Żabka")
show_tokens("My favourite shop is Żabka")
show_tokens('for i in range(10): print("hello")')
```

💡 **Notice:** common English words are usually **1 token**. Rare words, names and other languages get chopped into many pieces.
That's why AI used to struggle to count the letters in "strawberry": it never sees the letters, only the chunks!
Tokens also matter for **cost** (AI companies charge per token) and for the **context window** (how many tokens a model can read at once).

## 2️⃣ Vectors: the AI's map of meaning
Every token becomes a long list of numbers called a **vector**. Think of it as an **address on a map of meaning**.
Words with similar meanings live close together. "Close" is measured with **cosine similarity**: 1 = same direction (same meaning), 0 = unrelated.

```python
# @title 🔧 Load a small embedding model
from sentence_transformers import SentenceTransformer
embedder = SentenceTransformer("all-MiniLM-L6-v2")

vector = embedder.encode("cat")
print(f'The AI stores the word "cat" as a list of {len(vector)} numbers, like this:', np.round(vector[:5], 2), "...")
print("Too many numbers to imagine! So let's draw them as a map instead 👇")
```

```python
# @title 🗺️ Draw the map of meaning
WORD_GROUPS = {
    "animals":  ["cat", "dog", "lion", "tiger", "horse", "rabbit"],
    "food":     ["pizza", "pasta", "burger", "chocolate", "kebab", "pierogi"],
    "sports":   ["football", "basketball", "tennis", "swimming"],
    "tech":     ["computer", "router", "smartphone", "internet", "password", "hacker"],
}


def draw_word_map(groups):
    words = [w for ws in groups.values() for w in ws]
    vecs = embedder.encode(words, normalize_embeddings=True)
    centred = vecs - vecs.mean(axis=0)
    _, _, vt = np.linalg.svd(centred, full_matrices=False)
    xy = centred @ vt[:2].T
    plt.figure(figsize=(11, 8))
    i = 0
    for group, ws in groups.items():
        pts = xy[i:i + len(ws)]
        plt.scatter(pts[:, 0], pts[:, 1], s=120, label=group)
        for (x, y), w in zip(pts, ws):
            plt.annotate(w, (x, y), xytext=(6, 4), textcoords="offset points", fontsize=13)
        i += len(ws)
    plt.legend(fontsize=12)
    plt.title("The map of meaning: similar words live close together", fontsize=15)
    plt.xticks([]); plt.yticks([])
    plt.show()


draw_word_map(WORD_GROUPS)
```

```python
# @title 📐 How close are two meanings? (cosine similarity)
def similarity(a, b):
    va, vb = embedder.encode([a, b], normalize_embeddings=True)
    score = float(va @ vb)
    print(f"{score:5.2f}  {'█' * int(max(score, 0) * 30):30}  {a!r} vs {b!r}")


similarity("cat", "kitten")
similarity("cat", "dog")
similarity("cat", "router")
similarity("I failed my driving test", "I didn't pass my driving exam")
similarity("I failed my driving test", "I passed my driving test")
```

## 3️⃣ Context: which *Harry* do you mean?
1. "**Harry** caught the Golden Snitch on his broomstick."
2. "Prince **Harry** visited the King at Buckingham Palace."
3. "**Harry** Styles sang his new song on stage."

Same word, three different people! At the start, the token "Harry" gets the **same vector** in every sentence.
Then the AI mixes in information from the neighbouring words, and the vector *moves* on the map.

### 🔢 Before reading the neighbours: the AI starts out confused
Let's ask a real language model (**BERT**, made by Google). We give it **9 sentences: 3 about each Harry**.
First: what does "Harry" look like when the AI *starts* reading?

```python
# @title 🔧 Load BERT (a language model that is good at understanding context)
import re
from transformers import AutoTokenizer, AutoModel
bert_tok = AutoTokenizer.from_pretrained("bert-base-uncased")
bert = AutoModel.from_pretrained("bert-base-uncased", attn_implementation="eager")
bert.eval()

# The map shows the first ON_THE_MAP sentences of each group. The game compares with all of them: the extra ones put
# "Harry" in other positions, otherwise BERT matches the sentence pattern ("Harry" + verb) instead of the meaning.
# The map squeezes 768 numbers into 2, which can hide the groups: these first three per group stay apart on it.
HARRYS = {
    "wizard": ["Harry caught the Golden Snitch on his broomstick.",
               "Harry waved his wand and cast a spell on the troll.",
               "Harry and Ron ran to Hagrid's hut after their potions lesson.",
               "Hermione helped Harry with his homework at Hogwarts.",
               "Harry Potter fought Voldemort with his wand."],
    "prince": ["Prince Harry visited the King at Buckingham Palace.",
               "Prince Harry wore his royal uniform at the ceremony.",
               "The King's son Harry married Meghan at Windsor Castle.",
               "Harry and his brother William waved to the crowd from the palace balcony.",
               "Harry wrote a book about leaving the royal family."],
    "singer": ["Harry's new album went straight to number one in the music charts.",
               "Fans screamed when Harry sang his biggest hit.",
               "Harry went on a world tour and sold out every stadium.",
               "Harry Styles sang his new song on stage.",
               "Harry Styles played the guitar at his concert."],
}
COLORS_BY_GROUP = ["#e8453c", "#1a73e8", "#34a853", "#f5b400"]
SURENESS = 10
LAYERS = 4
ON_THE_MAP = 3


def word_in_context(sentence, word):
    """The vector BERT gives to `word` inside `sentence` (after it has read the whole sentence):
    the average of its last LAYERS layers."""
    found = re.search(rf"\b{re.escape(word)}(e?s)?\b", sentence, flags=re.IGNORECASE)
    if not found:
        raise ValueError(f'I can\'t find the word "{word}" in: "{sentence}". Check the spelling!')
    start, end = found.span()
    enc = bert_tok(sentence, return_tensors="pt", return_offsets_mapping=True)
    offsets = enc.pop("offset_mapping")[0].tolist()
    positions = [i for i, (s, e) in enumerate(offsets) if e > s and s < end and e > start]
    with torch.no_grad():
        layers = bert(**enc, output_hidden_states=True).hidden_states[-LAYERS:]
    hidden = torch.stack(layers).mean(dim=0)[0]
    return torch.nn.functional.normalize(hidden[positions].mean(dim=0), dim=0)


def on_the_map(examples):
    """The numbered sentences drawn on the map: the first ON_THE_MAP of each group, with the group's colour."""
    colors = {group: COLORS_BY_GROUP[k % len(COLORS_BY_GROUP)] for k, group in enumerate(examples)}
    return [(n, group, sentence, colors[group]) for n, (group, sentence) in
            enumerate(((g, s) for g, sentences in examples.items() for s in sentences[:ON_THE_MAP]), start=1)]


def same_start(word, examples):
    token_ids = bert_tok.convert_tokens_to_ids(bert_tok.tokenize(word))
    token_text = " + ".join(str(t) for t in token_ids)
    sentences = on_the_map(examples)
    for n, group, sentence, color in sentences:
        marked = html.escape(sentence).replace(
            word, f'<b style="background:#ffe08a;color:black;border-radius:5px;padding:0 5px">{word}</b>')
        display(HTML(f'<div style="font-size:19px;margin:5px 0"><b style="background:{color};color:white;'
                     f'border-radius:12px;padding:0 9px">{n}</b> {marked} &nbsp;→&nbsp; '
                     f'<b>{word}</b> = token <b>{token_text}</b></div>'))
    print(f'\n🤔 At the start, "{word}" is the SAME token (and the same vector) in all {len(sentences)} sentences.')
    print("   The AI can't tell them apart yet. The neighbouring words have to fix that!")


def examples_for(word, examples):
    if examples is not None:
        return examples
    if word.lower() in WORD_EXAMPLES:
        return WORD_EXAMPLES[word.lower()]
    print(f'😕 "{word}" isn\'t in the word list yet. Pick a word from the list, or add it to WORD_EXAMPLES!')
    return None


def meaning_map(word, examples=None):
    examples = examples_for(word, examples)
    if examples is None:
        return
    sentences = on_the_map(examples)
    vecs = torch.stack([word_in_context(sentence, word) for _, _, sentence, _ in sentences]).numpy()
    centred = vecs - vecs.mean(axis=0)
    xy = centred @ np.linalg.svd(centred, full_matrices=False)[2][:2].T
    plt.figure(figsize=(9, 6))
    for group in examples:
        rows = [i for i, (_, g, _, _) in enumerate(sentences) if g == group]
        points, color = xy[rows], sentences[rows[0]][3]
        plt.scatter(points[:, 0], points[:, 1], s=500, color=color, alpha=0.85)
        for i, (x, y) in zip(rows, points):
            plt.text(x, y, str(sentences[i][0]), ha="center", va="center", fontsize=13, fontweight="bold", color="white")
        plt.annotate(f"{word} = {group}", (points[:, 0].mean(), points[:, 1].max()), textcoords="offset points",
                     xytext=(0, 22), ha="center", fontsize=16, fontweight="bold", color=color)
    plt.margins(0.25)
    plt.xticks([])
    plt.yticks([])
    plt.title(f'After reading the neighbours: dot 1 is "{word}" in sentence 1, and so on', fontsize=15)
    plt.show()
    for n, group, sentence, _ in sentences:
        print(f"{n}. ({group}) {sentence}")


def which_meaning(sentence, word, examples=None):
    examples = examples_for(word, examples)
    if examples is None:
        return
    refs = {group: torch.stack([word_in_context(s, word) for s in sentences]) for group, sentences in examples.items()}
    # Every vector of the same word shares a big "this is the word Harry" part. Subtract it, so only the context is compared.
    shared = torch.cat(list(refs.values())).mean(dim=0)
    centre = lambda v: torch.nn.functional.normalize(v - shared, dim=-1)
    vec = centre(word_in_context(sentence, word))
    closeness = {group: centre(vecs).matmul(vec).mean().item() for group, vecs in refs.items()}
    scores = np.array(list(closeness.values()))
    sure = np.exp(SURENESS * (scores - scores.max()))
    sure = 100 * sure / sure.sum()
    best = max(closeness, key=closeness.get)
    print(f'❓ "{sentence}"')
    print(f'🤖 I think "{word}" here means: {best.upper()}!')
    fig, ax = plt.subplots(figsize=(8, 0.8 * len(examples) + 0.8))
    ax.barh([f"{word} = {g}" for g in examples], sure,
            color=["#34a853" if g == best else "#dadce0" for g in examples])
    for i, percent in enumerate(sure):
        ax.text(percent + 1.5, i, f"{percent:.0f}%", va="center", fontsize=14)
    ax.set_xlim(0, 115)
    ax.invert_yaxis()
    ax.set_xticks([])
    ax.tick_params(axis="y", labelsize=14, length=0)
    for side in ["top", "right", "bottom"]:
        ax.spines[side].set_visible(False)
    ax.set_title("How sure is the AI?", fontsize=14)
    plt.tight_layout()
    plt.show()


def which_harry(sentence):
    which_meaning(sentence, "Harry", HARRYS)
```

## 4️⃣ The giant guessing machine: predicting the next word
**GPT-2** (2019) is a small, older cousin of ChatGPT. It does just ONE thing: it gives **every possible next word a chance**.
It has 124 million **parameters** (numbers it learned during training). Today's big AIs have hundreds of billions!

```python
# @title 🔧 Load GPT-2
from transformers import AutoModelForCausalLM
gpt2_tok = AutoTokenizer.from_pretrained("gpt2")
gpt2 = AutoModelForCausalLM.from_pretrained("gpt2")
gpt2.eval()


def next_word_probabilities(text):
    ids = gpt2_tok(text, return_tensors="pt").input_ids
    with torch.no_grad():
        logits = gpt2(ids).logits[0, -1]
    return logits


def show_next_words(text, top=10):
    probs = torch.softmax(next_word_probabilities(text), dim=-1)
    p, idx = probs.topk(top)
    words = [repr(gpt2_tok.decode([int(i)])) for i in idx]
    plt.figure(figsize=(9, 5))
    plt.barh(words[::-1], (p * 100).numpy()[::-1], color="#7fb3ff")
    for y, v in enumerate((p * 100).numpy()[::-1]):
        plt.text(v + 0.5, y, f"{v:.1f}%", va="center", fontsize=12)
    plt.xlim(0, 100)
    plt.title(f'"{text} ..."', fontsize=14)
    plt.xlabel("probability (%)", fontsize=12)
    plt.yticks(fontsize=13)
    plt.tight_layout()
    plt.show()
```

```python
show_next_words("The capital of France is")
```

### ✍️ Writing one word at a time
To write a sentence, the model **picks one word, adds it to the text, and predicts again**, over and over.
It picks *randomly*, but likely words get picked more often, like rolling a weighted dice 🎲.

**Temperature** controls how adventurous it is:
- low (0.2): almost always picks the favourite word, so it's sensible but repetitive
- high (1.5): unlikely words get a real chance, so it's creative... or nonsense

```python
# @title 🔧 Text writer
TEMPERATURES = {0.2: ("🧊", "careful"), 0.8: ("🙂", "balanced"), 1.5: ("🔥", "wild")}


def pick_next_word(text, temperature):
    probs = torch.softmax(next_word_probabilities(text) / temperature, dim=-1)
    choice = torch.multinomial(probs, 1).item()
    return gpt2_tok.decode([choice]), probs[choice].item()


def write(text, tokens=15, temperature=0.8):
    print(f"🟦 Start: {text}\n")
    for step in range(1, tokens + 1):
        token, chance = pick_next_word(text, temperature)
        print(f"   step {step:2d}: picked {token!r:15} (chance: {chance * 100:4.1f}%)")
        text += token
    print(f"\n📖 Result: {text}")


def compare_temperatures(text, tokens=15, top=3):
    logits = next_word_probabilities(text)
    top_ids = logits.topk(top).indices
    words = [f'"{gpt2_tok.decode([int(i)]).strip()}"' for i in top_ids]
    labels = [f"any other word\n({len(logits) - top:,} of them)"] + words[::-1]
    colors = ["#bdc1c6"] + ["#7fb3ff"] * top
    fig, axes = plt.subplots(1, len(TEMPERATURES), figsize=(16, 4.8))
    for ax, (temperature, (_, name)) in zip(axes, TEMPERATURES.items()):
        p = (torch.softmax(logits / temperature, dim=-1)[top_ids] * 100).numpy()
        values = [100 - p.sum()] + list(p[::-1])
        ax.barh(labels, values, color=colors)
        for y, v in enumerate(values):
            ax.text(v + 1, y, "<1%" if 0 < v < 0.5 else f"{v:.0f}%", va="center", fontsize=13)
        ax.set_xlim(0, 115)
        ax.set_xticks([])
        ax.set_title(f"{name} (temperature {temperature})", fontsize=15)
        ax.tick_params(axis="y", labelsize=13, length=0)
        for side in ["top", "right", "bottom"]:
            ax.spines[side].set_visible(False)
    fig.suptitle(f'Which word comes after "{text} ..."?\n'
                 "Each bar is the chance that the word gets picked next. In each chart, the bars add up to 100%.",
                 fontsize=14)
    fig.tight_layout(rect=(0, 0, 1, 0.9))
    plt.show()
    print(f"🟦 Start: {text}\n")
    for temperature, (icon, name) in TEMPERATURES.items():
        ending = ""
        for _ in range(tokens):
            ending += pick_next_word(text + ending, temperature)[0]
        ending = " ".join(ending.split())
        print(f"{icon} {name} ({temperature}): ... {ending}\n")
```

```python
write("In the year 2050, the most important job skill will be", temperature=0.8)
```

```python
# ✏️ Same start, three temperatures. Run it a few times, then try your own start!
compare_temperatures("In the year 2050, the most important job skill will be")
```

### 🏋️ From GPT to ChatGPT
GPT-2 only did **pre-training** (predict the next token on lots of text), so it just continues text. ChatGPT-style assistants add two more stages:
**fine-tuning** on example conversations, and **RLHF** (reinforcement learning from human feedback: people rank answers, and the model is rewarded for the preferred ones).
Pre-training stops at a date, the **knowledge cutoff**. After that, the model knows nothing new unless we give it documents (that's RAG, in Lab 1).

---
## 🎯 The big idea
> **An LLM is a very large statistical machine that predicts the next word based on the context.**

1. Text → **tokens**
2. Tokens → **vectors** (positions on the map of meaning)
3. **Attention** mixes in context ("which Harry?")
4. Predict the **probability** of every possible next token
5. Pick one, add it, repeat!

It never "looks up" facts. It picks words that *sound* right. That's why it can sound confident and still be wrong (**hallucination**).
